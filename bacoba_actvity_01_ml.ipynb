```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
```

```python
iris = load_iris()
print(iris.feature_names)
```

```python
print(iris.target_names)
```

```python
df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

df["species"] = iris.target_names[iris.target]
df.head()
```

```python
print("Number of rows: ", df.shape[0])
print("number of columns ", df.shape[1])
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
print(df["species"].unique())
```

```python
print(df["species"].value_counts())
```

```python
plt.figure(figsize=(8,6))


plt.scatter(

    df["sepal length (cm)"],
    df["sepal width (cm)"],
)

plt.xlabel("Sepal length (cm)")
plt.ylabel("Sepal width (cm)")
plt.title("Sepal length vs Sepal width")

plt.show()
```

```python
plt.figure(figsize=(8,6))

for species in df["species"].unique():

  subset =df[df["species"]==species]

  plt.scatter(
      subset["petal length (cm)"],
       subset["petal width (cm)"],
      label=species
  )
  plt.xlabel("petal length (cm)")
plt.ylabel("petal width (cm)")
plt.title("Iris Species based on sepal Measurements")
plt.legend()

plt.show()


```

```python
x = iris.data
y = iris.target

X_train, X_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
print("Training observation:", len (X_train))
print("testing observation:", len (X_test))
```

```python
model = KNeighborsClassifier(n_neighbors=3)
```

```python
model.fit(X_train, y_train)
```

```python
y_pred=model.predict(X_test)
print(y_pred)
```

```python
print("Actual: ")
print(y_test)

print("\nPredicted: ")
print(y_pred)
```

```python
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy: ", accuracy)
```

```python
new_flower = [(6.0, 2.7, 5.1, 1.6)]
```

```python
prediction=model.predict(new_flower)

print(
    "predicted species: ",
    iris.target_names[prediction[0]]
)
```

```python
probabilities = model.predict_proba(new_flower)
print(probabilities)
```

```python
for species, probability in zip(
    iris.target_names,
    probabilities[0]
):
  print(
      f"{species}: {probability*100:.2f}%"
)
```

```python
k_values = [1, 3, 5, 7, 9]
accuracy_scores = []

for k in k_values:
  model = KNeighborsClassifier(n_neighbors=k)
  model.fit(X_train, y_train)
  y_pred = model.predict(X_test)
  accuracy = accuracy_score(y_test, y_pred)
  accuracy_scores.append(accuracy)
  print(f"k={k}, Accuracy = {accuracy:2f}")
```

```python
plt.figure(figsize=(8,5))
plt.plot(
    k_values,
    accuracy_scores,
    marker='o'
)
plt.xlabel("k values")
plt.ylabel("Accuracy")
plt.title("KNN Model Accuracy for Different k values")

plt.xticks(k_values)
plt.show()
```
